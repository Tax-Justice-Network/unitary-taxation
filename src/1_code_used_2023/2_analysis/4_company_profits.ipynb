This notebook aims to determine whether the rise in absolute profit shiftings is due to an increase in company profits, an increase in the percentage of profits shifted, or both.

```python
import pandas as pd
from config import YEAR_A, YEAR_B, PATH_PSO_A_CLEAN, PATH_PSO_B_CLEAN, PATH_MNC_PROFITS_B, PATH_MNC_PROFITS_A
```

```python
# Retrieve profits shiftings outwards for years A and B
pso_b = pd.read_excel(PATH_PSO_B_CLEAN)
pso_a = pd.read_excel(PATH_PSO_A_CLEAN)
pso_b
```

```python
# TODO Check Replicates values 2017 - Should sum up 4.7 trillion
companies_profit_b = pd.read_csv(PATH_MNC_PROFITS_B, delimiter='\s+') 

# Keep only columns of interest
companies_profit_b = companies_profit_b.loc[:,['iso3_d','iso3_o','profits']]
companies_profit_b
```

```python
# TODO Check Replicates values 2017 - Should sum up 4.7 trillion
companies_profit_a = pd.read_csv(PATH_MNC_PROFITS_A, delimiter='\s+') 

# Keep only columns of interest
companies_profit_a = companies_profit_a.loc[:,['iso3_d','iso3_o','profits']]
companies_profit_a
```

```python
# TODO Identify cause of wrong units for total profits - 1000 is a temporary fix
# Calculate total profits for years B and A
total_companies_profit_b = companies_profit_b['profits'].sum() / 1000 # TODO Delete / 1000 once we have the right data
total_companies_profit_a = companies_profit_a['profits'].sum() / 1000 # TODO Delete / 1000 once we have the right data
print(f'The total companies profit for {YEAR_B} is: ',f"{total_companies_profit_b:,.0f}", 'USD')
print(f'The total companies profit for {YEAR_A} is: ',f"{total_companies_profit_a:,.0f}", 'USD')
print(f'The increase in benefits between {YEAR_B} and {YEAR_A} is: ',f"{(total_companies_profit_a*100)/total_companies_profit_b - 100:,.2f}", '%')
```

```python
# Calculate profits shifted for years B and A
total_pso_b = pso_b['shifted_profits_outward'].sum()
total_pso_a = pso_a['shifted_profits_outward'].sum()
print(f'The total profits shifted outward for {YEAR_B} is:', f"{total_pso_b:,.0f}", 'USD')
print(f'The total profits shifted outward for {YEAR_A} is:', f"{total_pso_a:,.0f}", 'USD')
```

```python
# Calculate the proportion of profit that was shifted in year B in comparison to the total profits
share_profit_shifted_b = (total_pso_b*100)/total_companies_profit_b
print(f'The proportion of profit that was shifted in {YEAR_B} in comparison to the total profits is: ', f"{share_profit_shifted_b:,.2f}", '%')

# Calculate the proportion of profit that was shifted in year A in comparison to the total profits
share_profit_shifted_a = (total_pso_a*100)/total_companies_profit_a
print(f'The proportion of profit that was shifted in {YEAR_A} in comparison to the total profits is: ', f"{share_profit_shifted_a:,.2f}" , '%')
print(f'If the share of profit shifted in {YEAR_A} is higher than the share of profit shifted in {YEAR_B}, then corporations have shifted a higher proportion of their profits in {YEAR_A} than in {YEAR_B} \n')

# Calculate the proportional change in profits shifted between year B and A in comparison to the total profits
proportional_change_profit_shifted_a_b = ((share_profit_shifted_a - share_profit_shifted_b) / share_profit_shifted_b ) * 100
print(f'The proportional change in profits shifted between {YEAR_B} and {YEAR_A} in comparison to the total profits is: ', f"{proportional_change_profit_shifted_a_b:,.2f}", '%')
```
