## Filter for Norway from the Intermediate Datasets and export to Output and Specific Country Requests

Since we only need some basic CBRC data - and no data on profit misalignment, we can just filter for the Norway rows in the initial dataset downloaded from the OECD, and that we have cleaned.

```python
# Open the Intermediate Dataset
cbcr = pd.read_excel(f'{data_intermediate}/2_cbcr_second_cleaned_version_wide_format.xlsx')

# Filter for all rows where Norway is the parent jurisdiction
norway_parent = cbcr[cbcr['iso_parent'] == 'NOR']

# Filter for all rows where Norway is the parent jurisdiction
norway_partner = cbcr[cbcr['iso_partner'] == 'NOR']

# Export to Excel in output/specific_country_requests
norway_parent.to_excel('../output/specific_country_requests/norway_parent.xlsx', index=False)
norway_partner.to_excel('../output/specific_country_requests/norway_partner.xlsx', index=False)
```
