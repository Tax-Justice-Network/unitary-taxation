# The State of Tax Justice: Race to the Bottom. Comparison between CITs in 2016 and 2017.


## 0. Load packages

```python
# Packages
import pandas as pd
import numpy as np
import tjn_tools
from config import *

# Show columns and select data format
pd.set_option('display.max_columns', None)
pd.options.display.float_format = '{:,.2f}'.format
```

## Step 1. Import data


```python
cbcr_sample = pd.read_csv(f'{data_final}/cbcr_main_no_imputation_allsubgroupsonly.csv')

# Exclude country groups
cbcr_sample = cbcr_sample[~cbcr_sample['iso_partner'].isin(non_countries)]

cbcr_sample
```

## Step 2. Create samples with CIT 2016 and 2021

### Step 2.1. Create a sample with the CIT of 2016.

```python
# Keep cit_sample
cit_sample_2016 = cbcr_sample[cbcr_sample['year'] == 2016].copy()

# Keep iso_parent, year and cit
cit_sample_2016 = cit_sample_2016[['iso_partner', 'cit']]

# Drop duplicates
cit_sample_2016 = cit_sample_2016.drop_duplicates()

# Rename cit as cit_2016
cit_sample_2016 = cit_sample_2016.rename(columns={'cit': 'cit_2016'})

# Sort by iso_partner
cit_sample_2016 = cit_sample_2016.sort_values(by=['iso_partner'])

cit_sample_2016

```

### Step 2.2. Create a sample with the CIT of 2021.

```python
# Keep cit_sample
cit_sample_2021 = cbcr_sample[cbcr_sample['year'] == 2021].copy()

# Keep iso_parent, year and cit
cit_sample_2021 = cit_sample_2021[['iso_partner', 'cit']]

# Drop duplicates
cit_sample_2021 = cit_sample_2021.drop_duplicates()

# Rename cit as cit_2016
cit_sample_2021 = cit_sample_2021.rename(columns={'cit': 'cit_2021'})

# Sort by iso_partner
cit_sample_2021 = cit_sample_2021.sort_values(by=['iso_partner'])

cit_sample_2021
```

## Step 3. Merge the datasets and generate the difference between CITs

```python
# Merge cit_sample_2016 and cit_sample_2021
comparison_cits = pd.merge(cit_sample_2016, cit_sample_2021, how='inner', on='iso_partner')

# Sort by iso_partner
comparison_cits = comparison_cits.sort_values(by=['iso_partner'])

# Generate decrease in CIT by subtracting cit_2016 from cit
comparison_cits['decrease_cit'] = comparison_cits['cit_2021'] - comparison_cits['cit_2016']

comparison_cits
```

## Step 4. Comparative Results.

### Step 4.1 Number of countries that decreased their CIT, and average decrease

```python
# Filter Negative Values
decrease_cit = comparison_cits[comparison_cits['decrease_cit'] < 0]

# Estimate the Average decrease in CIT
decrease_cit['decrease_cit'].mean()
print('The average decrease in CIT is:', decrease_cit['decrease_cit'].mean())

# How many values does the dataset have?
decrease_cit['decrease_cit'].count()
print('This is the number of countries that decreased their CIT:', decrease_cit['decrease_cit'].count())

decrease_cit
```

### Step 4.2. Number of countries that increased their CIT, and average decrease

```python
# Filter Positive Values
increase_cit = comparison_cits[comparison_cits['decrease_cit'] > 0]

# Estimate the Average decrease in CIT
increase_cit['decrease_cit'].mean()
print('The average decrease in CIT is:', increase_cit['decrease_cit'].mean())

# How many values does the dataset have?
increase_cit['decrease_cit'].count()
print('This is the number of countries that decreased their CIT:', increase_cit['decrease_cit'].count())

increase_cit
```

### Step 4.3. Number of countries that changed their CIT, and average change

```python
# Filter 0 Values
cit_unchanged = comparison_cits[comparison_cits['decrease_cit'] != 0]

# Estimate the Average decrease in CIT
cit_unchanged['decrease_cit'].mean()
print('The average decrease in CIT is:', cit_unchanged['decrease_cit'].mean())

# How many values does the dataset have?
cit_unchanged['decrease_cit'].count()
print('This is the number of countries that changed their CIT:', cit_unchanged['decrease_cit'].count(), 'values')

cit_unchanged
```
