This notebook will preprocess the input raw datasets `2018_tax_avoidance_sotj_table.xlsx` and `sotj_table_3_english_2021.xlsx` to produce in the intermediary folder profit_shiftings_outward_2022.xlsx and profit_shiftings_outward_2021.xlsx dataset, respectively.

```python
import pandas as pd
import tjn_tools
```

### Profit Shiftings Outward 2022

```python
pso_2022 = pd.read_excel('../data/raw/2018_tax_avoidance_sotj_table.xlsx')
pso_2022.head()
```

```python
# Keep only columns of interest
pso_2022 = pso_2022.loc[:,['Name', 'Profit loss (M)']]
# Add iso3 column
pso_2022.insert(0, "iso3", pso_2022["Name"].apply(tjn_tools.name_to_iso3, print_failure=True))
# Drop regions rows
pso_2022 = pso_2022.loc[pso_2022['iso3'].notnull()]
# Convert negative Profit loss (M) numbers to 0 - To only consider outward profit shiftings
# and keep track of all initial countries available
pso_2022.loc[pso_2022['Profit loss (M)'] <=0,'Profit loss (M)'] = 0
# Group profit shiftings by country
pso_2022 = pso_2022.groupby(by=['iso3', 'Name']).sum().reset_index()
# Rename columns
pso_2022.rename(columns={'Name':'Country','Profit loss (M)':'Shifted profits outward (USD million)'}, inplace=True)
pso_2022
```

```python
# Check total profit shifting is equal to 1.58 trillion
pso_2022['Shifted profits outward (USD million)'].sum()
```

```python
# Save dataset in intermediate folder
pso_2022.to_excel('../data/intermediate/profits_shifting_outward_2022.xlsx', index=False)
```
