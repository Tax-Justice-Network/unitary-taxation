This notebook will check whether the increase in profit shiftings stem from the country coverage

```python
import pandas as pd
import tjn_tools
```

```python
# Retrieve profits shiftings outwards 2021 (2017 data) and 2022 (2018 data)
pso_2021 = pd.read_excel('../data/intermediate/profits_shifting_outward_2021.xlsx')
pso_2022 = pd.read_excel('../data/intermediate/profits_shifting_outward_2022.xlsx')
pso_2021
```

```python
num_countries_covered_2021 = len(pso_2021['iso3'].unique().tolist())
num_countries_covered_2022 = len(pso_2022['iso3'].unique().tolist())
print('Countries covered in 2021: ', num_countries_covered_2021)
print('Countries covered in 2022: ', num_countries_covered_2022)
```

```python
countries_covered_2021 = pso_2021['iso3'].unique().tolist()
countries_covered_2022 = pso_2022['iso3'].unique().tolist()

common_elements = []
elements_only_in_2021 = []
elements_only_in_2022 = []

for element in countries_covered_2021:
    if element in countries_covered_2022:
        common_elements.append(element)
    else:
        elements_only_in_2021.append(element)

for element in countries_covered_2022:
    if element not in countries_covered_2021:
        elements_only_in_2022.append(element)

print("Common elements:", common_elements)
print("Number of countries only in 2021:", len(elements_only_in_2021))
print("Number of countries only in 2022:", len(elements_only_in_2022))
print("Countries only in 2021:", [ tjn_tools.data_processing.iso3_to_name(iso3) for iso3 in elements_only_in_2021])
print("Countries only in 2022:", [ tjn_tools.data_processing.iso3_to_name(iso3) for iso3 in elements_only_in_2022])
```

```python
# Calculate profit shiftings from countries added in 2021
pso_2021.loc[pso_2021['iso3'].isin(elements_only_in_2021),'Shifted profits outward (USD million)'].sum()
```

```python
# Calculate profit shiftings from countries added in 2022
pso_2022.loc[pso_2022['iso3'].isin(elements_only_in_2022),'Shifted profits outward (USD million)'].sum()
```
