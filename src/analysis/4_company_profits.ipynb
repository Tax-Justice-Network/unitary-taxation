```python
import pandas as pd
import tjn_tools
```

```python
# Retrieve profits shiftings outwards 2021 (2017 data) and 2022 (2018 data)
pso_2021 = pd.read_excel('../data/intermediate/profits_shifting_outward_2021.xlsx')
pso_2022 = pd.read_excel('../data/intermediate/profits_shifting_outward_2022.xlsx')
pso_2021
```

```python
# Check Replicates values 2017 - Should sum up 4.7 billion
companies_profit_2017 = pd.read_csv('../data/raw/2017_replicates.csv', delimiter='\s+') 

# Keep only columns of interest
companies_profit_2017 = companies_profit_2017.loc[:,['iso3_d','iso3_o','profits']]
companies_profit_2017
```

```python
# Check Replicates values 2017 - Should sum up 4.7 billion
companies_profit_2018 = pd.read_csv('../data/raw/2018_replicates.csv', delimiter='\s+') 

# Keep only columns of interest
companies_profit_2018 = companies_profit_2018.loc[:,['iso3_d','iso3_o','profits']]
companies_profit_2018
```

```python
total_companies_profit_2017 = companies_profit_2017['profits'].sum()
total_companies_profit_2018 = companies_profit_2018['profits'].sum()
print('The total companies profit for 2017 is: ', total_companies_profit_2017)
print('The total companies profit for 2018 is: ', total_companies_profit_2018)
print(' The increase in benefits between 2017 and 2018 is: ',(total_companies_profit_2018*100)/total_companies_profit_2017 - 100 )
```
