## Concatenate Country Results

```python
# Concatenate Final Results
sotj_countries_2016 = pd.read_csv('../output/tables/2024/Final_Full_CBCR_Datasets/SOTJ_sample_countries_2016.csv')
sotj_countries_2017 = pd.read_csv('../output/tables/2024/Final_Full_CBCR_Datasets/SOTJ_sample_countries_2017.csv')
sotj_countries_2018 = pd.read_csv('../output/tables/2024/Final_Full_CBCR_Datasets/SOTJ_sample_countries_2018.csv')
sotj_countries_2019 = pd.read_csv('../output/tables/2024/Final_Full_CBCR_Datasets/SOTJ_sample_countries_2019.csv')
sotj_countries_2020 = pd.read_csv('../output/tables/2024/Final_Full_CBCR_Datasets/SOTJ_sample_countries_2020.csv')
sotj_countries_2021 = pd.read_csv('../output/tables/2024/Final_Full_CBCR_Datasets/SOTJ_sample_countries_2021.csv')

# Add a column with the value 2016 to sotj_countries_2016
sotj_countries_2016['year'] = 2016
sotj_countries_2017['year'] = 2017
sotj_countries_2018['year'] = 2018
sotj_countries_2019['year'] = 2019
sotj_countries_2020['year'] = 2020
sotj_countries_2021['year'] = 2021


#Concatenate all of the above
sotj_countries = pd.concat([sotj_countries_2016, sotj_countries_2017, sotj_countries_2018, sotj_countries_2019, sotj_countries_2020, sotj_countries_2021])

# Sort by iso_partner and then year
sotj_countries = sotj_countries.sort_values(by=['iso_partner', 'year'])

# Order columns first iso_partner, then year, and then the rest

# Order columns first iso_partner, then year, and then the rest
cols = ['iso_partner', 'year'] + [col for col in sotj_countries.columns if col not in ['iso_partner', 'year']]
sotj_countries = sotj_countries[cols]

sotj_countries
```

## Do it just for those with positive values


```python
cbcr_sample_positives = pd.read_csv('../data/final/2024/cbcr_main_no_imputation_allsubgroupsonly.csv')

# Filter if profit_loss_before_income_tax is positive
cbcr_sample_positives = cbcr_sample_positives[cbcr_sample_positives['profit_loss_before_income_tax_corrected'] > 0]

cbcr_sample_positives
```

```python
#Sum the total income_tax_accrued_current_year and the profit_loss_before_income_tax_corrected by years
etr_calculation_positives = cbcr_sample_positives.groupby('year')[['income_tax_accrued_current_year', 'profit_loss_before_income_tax_corrected']].sum()

# Calculate the effective tax rate
etr_calculation_positives['etr'] = etr_calculation_positives['income_tax_accrued_current_year'] / etr_calculation_positives['profit_loss_before_income_tax_corrected']

# Generate a new column called increased_etr = etr + 0.01
etr_calculation_positives['increased_etr'] = etr_calculation_positives['etr'] + 0.01

# Generate new column called new tax_revenue = increased_etr * income_tax_accrued_current_year
etr_calculation_positives['new_tax_revenue'] = etr_calculation_positives['increased_etr'] * etr_calculation_positives['profit_loss_before_income_tax_corrected']

# Generate a new column called extra revenue collected = new_tax_revenue - income_tax_accrued_current_year
etr_calculation_positives['extra_revenue_collected'] = etr_calculation_positives['new_tax_revenue'] - etr_calculation_positives['income_tax_accrued_current_year']

etr_calculation_positives
```
