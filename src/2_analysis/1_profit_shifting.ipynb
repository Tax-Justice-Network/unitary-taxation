This notebook will preprocess the input raw datasets of a year named reference and its previous year to produce in the intermediary folder profit_shiftings_outward_{YEAR_A}.xlsx and profit_shiftings_outward_{YEAR_B}.xlsx dataset, respectively.

### Imports

```python
import pandas as pd
import tjn_tools
from config import YEAR_A, YEAR_B, PATH_PSO_A, PATH_PSO_A_CLEAN, PATH_PSO_B, PATH_PSO_B_CLEAN
```

### Profit Shiftings Outward Reference Year

```python
pso_a = pd.read_excel(PATH_PSO_A)
pso_a.head()
```

```python
# Keep only columns of interest
pso_a = pso_a.loc[:,['Name', 'Profit loss (M)']]
# Add iso3 column
pso_a.insert(0, "iso3", pso_a["Name"].apply(tjn_tools.name_to_iso3, print_failure=True))
# Drop regions rows
pso_a = pso_a.loc[pso_a['iso3'].notnull()]
# Convert negative Profit loss (M) numbers to 0 - To only consider outward profit shiftings
# and keep track of all initial countries available
pso_a.loc[pso_a['Profit loss (M)'] <=0,'Profit loss (M)'] = 0
# Group profit shiftings by country
pso_a = pso_a.groupby(by=['iso3', 'Name']).sum().reset_index()
# Convert units to USD
pso_a['shifted_profits_outward'] = pso_a['Profit loss (M)'] * 1000000
# Rename columns and change units
pso_a.drop(columns=['Profit loss (M)'], inplace=True)
pso_a.rename(columns={'Name':'country'}, inplace=True)
pso_a
```

```python
# Check total profit shifting is equal to 1.58 trillion
print(f'Total profit shifting for {YEAR_A} is:', f"{pso_a['shifted_profits_outward'].sum():,.0f}", 'USD')
```

```python
# Save dataset in intermediate folder
pso_a.to_excel(PATH_PSO_A_CLEAN, index=False)
```
