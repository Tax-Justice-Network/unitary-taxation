This notebook will check whether the increase in profit shiftings stem from the country coverage.

```python
# Imports
import pandas as pd
import tjn_tools
from config import YEAR_A, YEAR_B, PATH_PSO_A_CLEAN, PATH_PSO_B_CLEAN
```

```python
# Retrieve profits shiftings outwards for years A and B
pso_b = pd.read_excel(PATH_PSO_B_CLEAN)
pso_a = pd.read_excel(PATH_PSO_A_CLEAN)
pso_b
```

```python
num_countries_covered_b = len(pso_b['iso3'].unique().tolist())
num_countries_covered_a = len(pso_a['iso3'].unique().tolist())
print(f'Countries covered in {YEAR_B} data: ', num_countries_covered_b)
print(f'Countries covered in {YEAR_A} data: ', num_countries_covered_a)
```

```python
countries_covered_b = pso_b['iso3'].unique().tolist()
countries_covered_a = pso_a['iso3'].unique().tolist()

common_elements = []
elements_only_in_b = []
elements_only_in_a = []

for element in countries_covered_b:
    if element in countries_covered_a:
        common_elements.append(element)
    else:
        elements_only_in_b.append(element)

for element in countries_covered_a:
    if element not in countries_covered_b:
        elements_only_in_a.append(element)

print("Common elements:", common_elements)
print(f"Number of countries only in {YEAR_B}:", len(elements_only_in_b))
print(f"Number of countries only in {YEAR_A}:", len(elements_only_in_a))
print(f"Countries only in {YEAR_B}:", [ tjn_tools.data_processing.iso3_to_name(iso3) for iso3 in elements_only_in_b])
print(f"Countries only in {YEAR_A}:", [ tjn_tools.data_processing.iso3_to_name(iso3) for iso3 in elements_only_in_a])
```

```python
# Calculate profit shiftings from countries added in YEAR_B
print(f'Profit shiftings from countries exclusively included in {YEAR_B} but not in {YEAR_A}:', f"{pso_b.loc[pso_b['iso3'].isin(elements_only_in_b),'shifted_profits_outward'].sum():,.0f}", 'USD')
```

```python
# Calculate profit shiftings from countries added in year A
print(f'Profit shiftings from countries exclusively included in {YEAR_A}:', f"{pso_a.loc[pso_a['iso3'].isin(elements_only_in_a),'shifted_profits_outward'].sum():,.0f}", 'USD')
```
