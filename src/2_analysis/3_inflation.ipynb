This notebook will check whether the increase in profit shiftings stem from the inflation.

```python
import pandas as pd
import tjn_tools
from config import YEAR_A, YEAR_B, PATH_PSO_A_CLEAN, PATH_PSO_B_CLEAN, PATH_INFLATION_WB
```

```python
# Retrieve profits shiftings outwards for Year A and B
pso_b = pd.read_excel(PATH_PSO_B_CLEAN)
pso_a = pd.read_excel(PATH_PSO_A_CLEAN)
pso_b
```

```python
# Retrieve inflation data
inflation = pd.read_csv(PATH_INFLATION_WB)
# Retrieve inflation of USA for Year A
inflation_usa_year_a = inflation.loc[(inflation['iso3']=='USA') & (inflation['year']==YEAR_A),'inflation'].values[0]
print(f'Inflation usa {YEAR_A}:', f"{inflation_usa_year_a:,.3f}", '%')
```

```python
# Calculate adjusted Year A data
pso_a_adjusted = pso_a.copy()
pso_a_adjusted.insert(3, 'adjusted_shifted_profits_outward', pso_a_adjusted['shifted_profits_outward'] * (1-((inflation_usa_year_a/100))))
pso_a_adjusted
```

```python
# Compare profit shifting between Year A data and adjusted Year A data
profits_shifting_a = pso_a_adjusted['shifted_profits_outward'].sum()
profits_shifting_a_adjusted = int(pso_a_adjusted['adjusted_shifted_profits_outward'].sum())

print(f'Profits shifted in {YEAR_A}:', f"{profits_shifting_a:,.0f}" , 'USD')
print(f'Profits shifted in {YEAR_A} with inflation adjustment:',  f"{profits_shifting_a_adjusted:,.0f}" , 'USD')
print(f'Increase in shifted in {YEAR_A} explained by inflation:',f"{profits_shifting_a - profits_shifting_a_adjusted:,.0f}", 'USD')
```
