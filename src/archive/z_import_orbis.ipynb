```python
import pandas as pd
```

```python
# Define the file path
file_path = r"D:/Orbis_raw/Ownership histo June text/Links_2016.txt"

# Create an empty DataFrame to store the processed data
all_chunks = pd.DataFrame()

# Process the file in chunks and remove duplicates for each chunk
chunksize = 1000  # Adjust chunk size based on your memory and performance needs

for chunk in pd.read_csv(file_path, sep='\t', encoding='ISO-8859-1', usecols=['GUO 50', 'GUO 50c'], chunksize=chunksize):
    # Remove duplicates within the chunk
    chunk = chunk.drop_duplicates()
    
    # Append the processed chunk to the final DataFrame
    all_chunks = pd.concat([all_chunks, chunk])

# Optionally, remove duplicates across all chunks in the final DataFrame
all_chunks = all_chunks.drop_duplicates()

# Display the resulting DataFrame
print(all_chunks)
```
