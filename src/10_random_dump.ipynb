### This was the first attempt at scaling the data 

These are the ratios based on (total foreign profits of full reporting countries and non-full reporting countries)/total foreign profits of full reporting countries

```python
# Initialize a list to store the new aggregate results
results_sample_fp_scaled = []

# Define the scaling values for each year
scaling_values = {
    2016: 2031184 / 926079, 
    2017: 4034180 / 1606588,
    2018: 4755417 / 1813498,
    2019: 4797061 / 1920169,
    2020: 3130374 / 1156379,
    2021: 6069250 / 2429281
}

# Loop through each year and process the corresponding file
for year in range(2016, 2022):
    # Open csv file
    sotj_f = pd.read_csv(f'{output_tables}/Sample_with_original_data_only_foreign_profits/three_factor_f_sample_{year}.csv')

    # Get the scaling value for the current year
    scaling_value = scaling_values[year]

    # Multiply negative_misalignment, tax_revenue_loss, positive_misalignment, tax_revenue_gain, tax_revenue_loss_caused_usd by scaling value
    sotj_f['negative_misalignment'] = sotj_f['negative_misalignment'] * scaling_value
    sotj_f['tax_revenue_loss'] = sotj_f['tax_revenue_loss'] * scaling_value
    sotj_f['positive_misalignment'] = sotj_f['positive_misalignment'] * scaling_value
    sotj_f['tax_revenue_gain'] = sotj_f['tax_revenue_gain'] * scaling_value
    sotj_f['tax_revenue_loss_caused_usd'] = sotj_f['tax_revenue_loss_caused_usd'] * scaling_value

    # Calculate the totals needed as denominators for the new shares and proportions
    total_tax_revenue_loss = sotj_f['tax_revenue_loss'].sum()
    total_positive_misalignment = sotj_f['positive_misalignment'].sum()
    total_negative_misalignment = sotj_f['negative_misalignment'].sum()
    total_profits = sotj_f['reported_profit'].sum()
    misaligned_of_total_profits = total_positive_misalignment / total_profits
    total_tax_revenue_gain = sotj_f['tax_revenue_gain'].sum()
    average_tax_revenue_loss_pct_of_gvt_health_expenditure = sotj_f['tax_revenue_loss_pct_of_gvt_health_expenditure'].mean()
    average_tax_revenue_loss_pct_of_total_tax_revenues = sotj_f['tax_revenue_loss_pct_of_total_tax_revenues'].mean()


    # Obtain the new shares and proportions
    sotj_f['tax_revenue_loss_suffered_pct_of_total'] = sotj_f['tax_revenue_loss'] / total_tax_revenue_loss
    sotj_f['tax_revenue_loss_pct_of_gvt_health_expenditure'] = np.where(
        sotj_f['gvt_health_expenditure'] == 0, 
        np.nan, 
        sotj_f['tax_revenue_loss'] / (sotj_f['gvt_health_expenditure'] / 1e6)
    )
    sotj_f['tax_revenue_loss_pct_of_total_tax_revenues'] = np.where(
        sotj_f['tax_revenue_current_usd'] == 0, 
        np.nan, 
        sotj_f['tax_revenue_loss'] / (sotj_f['tax_revenue_current_usd'] / 1e6)
    )
    sotj_f['tax_revenue_loss_caused_usd'] = sotj_f['tax_revenue_loss_caused_pct_of_total'] * total_tax_revenue_loss
    sotj_f['tax_revenue_loss_caused_pct_of_total'] = sotj_f['positive_misalignment'] / total_positive_misalignment

    # Select the relevant columns
    sotj_f = sotj_f[['iso_partner', 'partner_jurisdiction', 'negative_misalignment',
                     'tax_revenue_loss', 'tax_revenue_loss_suffered_pct_of_total', 'tax_revenue_loss_pct_of_gvt_health_expenditure',
                     'tax_revenue_loss_pct_of_total_tax_revenues', 'positive_misalignment', 'tax_revenue_gain', 
                     'tax_revenue_loss_caused_usd', 'tax_revenue_loss_caused_pct_of_total', 'etr_average_corrected', 'cit',
                     'region_tjn', 'ukt', 'oecd', 'oecd_oct', 'nld_oct']]

    # Save the scaled results to a new file
    sotj_f.to_csv(f'{output_tables}/Sample_with_origina_data_only_fp_scaled/three_factor_f_scaled_sample_{year}.csv', index=False) # CHANGE FILE NAME HERE, DEPENDING ON FORMULA USE

    # Append aggregate results to the list
    results_sample_fp_scaled.append({
        'year': year,
        'total_positive_misalignment': total_positive_misalignment,
        'total_negative_misalignment': total_negative_misalignment,
        'total_profits': total_profits,
        'misaligned_of_total_profits': misaligned_of_total_profits,
        'total_tax_revenue_loss': total_tax_revenue_loss,
        'total_tax_revenue_gain': total_tax_revenue_gain,
        'average_tax_revenue_loss_pct_of_gvt_health_expenditure': average_tax_revenue_loss_pct_of_gvt_health_expenditure,
        'average_tax_revenue_loss_pct_of_total_tax_revenues': average_tax_revenue_loss_pct_of_total_tax_revenues
    })

# Convert aggregate results to a DataFrame
results_sample_fp_scaled_df = pd.DataFrame(results_sample_fp_scaled)

# Save the aggregated results to a CSV or Excel file
results_sample_fp_scaled_df.to_csv(f'{output_tables}/Sample_with_origina_data_only_fp_scaled/three_factor_f_scaled_sample_aggregate_results.csv', index=False)  # CHANGE FILE NAME HERE, DEPENDING ON FORMULA USE
```
