```python
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.ticker as ticker
```

```python
# Load the CSV files
sotj = pd.read_csv('../output/tables/2024/SOTJ_aggregate_results.csv')
ccctb = pd.read_csv('../output/tables/2024/CCCTB_aggregate_results.csv')
doubleweightedsales = pd.read_csv('../output/tables/2024/Double-weighted-sales_aggregate_results.csv')
```

```python
plt.figure(figsize=(10, 6))
plt.plot(sotj['year'], sotj['total_positive_misalignment'], label='SOTJ Formula')
plt.plot(ccctb['year'], ccctb['total_positive_misalignment'], label='CCCTB Formula')
plt.plot(doubleweightedsales['year'], doubleweightedsales['total_positive_misalignment'], label='Double-Weighted Sales Formula')

plt.title('Total Misalignment Over the Years')
plt.xlabel('Year')
plt.ylabel('Total Misalignment (in million USD)')
plt.ylim(0, )
plt.legend()
plt.gca().yaxis.set_major_formatter(ticker.FuncFormatter(lambda x, pos: f'{x:,.0f}'))
plt.grid(True)
plt.show()
```

```python
plt.figure(figsize=(10, 6))
plt.plot(sotj['year'], sotj['total_tax_revenue_loss'], label='SOTJ Formula')
plt.plot(ccctb['year'], ccctb['total_tax_revenue_loss'], label='CCCTB Formula')
plt.plot(doubleweightedsales['year'], doubleweightedsales['total_tax_revenue_loss'], label='Double-Weighted Sales Formula')

plt.title('Total Tax Revenue Loss Over the Years')
plt.xlabel('Year')
plt.ylabel('Total Tax Revenue Loss (in million USD)')
plt.ylim(0, )
plt.legend()
plt.gca().yaxis.set_major_formatter(ticker.FuncFormatter(lambda x, pos: f'{x:,.0f}'))
plt.grid(True)
plt.show()
```

```python
plt.figure(figsize=(10, 6))
plt.plot(sotj['year'], sotj['total_tax_revenue_gain'], label='SOTJ Formula')
plt.plot(ccctb['year'], ccctb['total_tax_revenue_gain'], label='CCCTB Formula')
plt.plot(doubleweightedsales['year'], doubleweightedsales['total_tax_revenue_gain'], label='Double-Weighted Sales Formula')

plt.title('Total Tax Revenue Gain Over the Years')
plt.xlabel('Year')
plt.ylabel('Total Tax Revenue Gain (in million USD)')
plt.ylim(0, 300000)
plt.legend()
plt.gca().yaxis.set_major_formatter(ticker.FuncFormatter(lambda x, pos: f'{x:,.0f}'))
plt.grid(True)
plt.show()
```

```python
plt.figure(figsize=(10, 6))
plt.plot(sotj['year'], sotj['average_tax_revenue_loss_pct_of_gvt_health_expenditure'], label='SOTJ Formula')
plt.plot(ccctb['year'], ccctb['average_tax_revenue_loss_pct_of_gvt_health_expenditure'], label='CCCTB Formula')
plt.plot(doubleweightedsales['year'], doubleweightedsales['average_tax_revenue_loss_pct_of_gvt_health_expenditure'], label='Double-Weighted Sales Formula')

plt.title('Tax Revenue Loss as Percentage of Government Health Expenditure, Average')
plt.xlabel('Year')
plt.ylabel('Tax Revenue Loss as Percentage of Government Health Expenditure')
plt.ylim(0, )
plt.legend()
plt.grid(True)
plt.show()
```
