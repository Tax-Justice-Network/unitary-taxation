```python
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.ticker as ticker
```

### 1. Figures for sample with full reporting

```python
# Load the CSV files
sotj = pd.read_csv('../output/tables/2024/Sample_with_original_data_only/SOTJ_sample_aggregate_results.csv')
ccctb = pd.read_csv('../output/tables/2024/Sample_with_original_data_only/CCCTB_sample_aggregate_results.csv')
doubleweightedsales = pd.read_csv('../output/tables/2024/Sample_with_original_data_only/double_weighted_sales_sample_aggregate_results.csv')
canadian = pd.read_csv('../output/tables/2024/Sample_with_original_data_only/canadian_formula_sample_aggregate_results.csv')
threefactor = pd.read_csv('../output/tables/2024/Sample_with_original_data_only/three_factor_sample_aggregate_results.csv')
```

```python
plt.figure(figsize=(10, 6))
plt.plot(sotj['year'], sotj['total_positive_misalignment'], label='SOTJ Formula')
plt.plot(ccctb['year'], ccctb['total_positive_misalignment'], label='CCCTB Formula')
plt.plot(doubleweightedsales['year'], doubleweightedsales['total_positive_misalignment'], label='Double-Weighted Sales Formula')
plt.plot(canadian['year'], canadian['total_positive_misalignment'], label='Canadian Formula')
plt.plot(threefactor['year'], threefactor['total_positive_misalignment'], label='Three-Factor Formula')

plt.title('Total Misalignment Over the Years, based on full reporting countries only')
plt.xlabel('Year')
plt.ylabel('Total Misalignment (in million USD)')
plt.ylim(0, )
plt.legend()
plt.gca().yaxis.set_major_formatter(ticker.FuncFormatter(lambda x, pos: f'{x:,.0f}'))
plt.grid(True)
plt.show()
```

```python
plt.figure(figsize=(10, 6))
plt.plot(sotj['year'], sotj['misaligned_of_total_profits'] * 100, label='SOTJ Formula')
plt.plot(ccctb['year'], ccctb['misaligned_of_total_profits'] * 100, label='CCCTB Formula')
plt.plot(doubleweightedsales['year'], doubleweightedsales['misaligned_of_total_profits'] * 100, label='Double-Weighted Sales Formula')
plt.plot(canadian['year'], canadian['misaligned_of_total_profits'] * 100, label='Canadian Formula')
plt.plot(threefactor['year'], threefactor['misaligned_of_total_profits'] * 100, label='Three-Factor Formula')

plt.title('Fraction of Profit Shifted Over the Years, based on full reporting countries only')
plt.xlabel('Year')
plt.ylabel('Fraction of profit shifted (in %)')
plt.ylim(0, 30)  # Set y-axis limits to 0 to 100%
plt.legend()
plt.gca().yaxis.set_major_formatter(ticker.FuncFormatter(lambda x, pos: f'{int(x)}%'))  # Format y-axis as percentage
plt.grid(True)
plt.show()
```

```python
plt.figure(figsize=(10, 6))
plt.plot(sotj['year'], sotj['total_tax_revenue_loss'], label='SOTJ Formula')
plt.plot(ccctb['year'], ccctb['total_tax_revenue_loss'], label='CCCTB Formula')
plt.plot(doubleweightedsales['year'], doubleweightedsales['total_tax_revenue_loss'], label='Double-Weighted Sales Formula')
plt.plot(canadian['year'], canadian['total_tax_revenue_loss'], label='Canadian Formula')
plt.plot(threefactor['year'], threefactor['total_tax_revenue_loss'], label='Three-Factor Formula')

plt.title('Total Tax Revenue Loss Over the Years, based on full reporting countries only')
plt.xlabel('Year')
plt.ylabel('Total Tax Revenue Loss (in million USD)')
plt.ylim(0, )
plt.legend()
plt.gca().yaxis.set_major_formatter(ticker.FuncFormatter(lambda x, pos: f'{x:,.0f}'))
plt.grid(True)
plt.show()
```

```python
plt.figure(figsize=(10, 6))
plt.plot(sotj['year'], sotj['total_tax_revenue_gain'], label='SOTJ Formula')
plt.plot(ccctb['year'], ccctb['total_tax_revenue_gain'], label='CCCTB Formula')
plt.plot(doubleweightedsales['year'], doubleweightedsales['total_tax_revenue_gain'], label='Double-Weighted Sales Formula')
plt.plot(canadian['year'], canadian['total_tax_revenue_gain'], label='Canadian Formula')
plt.plot(threefactor['year'], threefactor['total_tax_revenue_gain'], label='Three-Factor Formula')

plt.title('Total Tax Revenue Gain Over the Years, based on full reporting countries only')
plt.xlabel('Year')
plt.ylabel('Total Tax Revenue Gain (in million USD)')
plt.ylim(0, )
plt.legend()
plt.gca().yaxis.set_major_formatter(ticker.FuncFormatter(lambda x, pos: f'{x:,.0f}'))
plt.grid(True)
plt.show()
```

### 3. Figures for global sample with imputed values

```python
# Load the CSV files
sotj = pd.read_csv('../output/tables/2024/Global_sample_with_imputed_data/One_sample/SOTJ_global_sample_aggregate_results.csv')
sotj = sotj[sotj['year'] != 2016]
ccctb = pd.read_csv('../output/tables/2024/Global_sample_with_imputed_data/One_sample/CCCTB_global_sample_aggregate_results.csv')
ccctb = ccctb[ccctb['year'] != 2016]
doubleweightedsales = pd.read_csv('../output/tables/2024/Global_sample_with_imputed_data/One_sample/double_weighted_sales_global_sample_aggregate_results.csv')
doubleweightedsales = doubleweightedsales[doubleweightedsales['year'] != 2016]
canadian = pd.read_csv('../output/tables/2024/Global_sample_with_imputed_data/One_sample/canadian_formula_global_sample_aggregate_results.csv')
canadian = canadian[canadian['year'] != 2016]
threefactor = pd.read_csv('../output/tables/2024/Global_sample_with_imputed_data/One_sample/three_factor_global_sample_aggregate_results.csv')
threefactor = threefactor[threefactor['year'] != 2016]
```

```python
plt.figure(figsize=(10, 6))
plt.plot(sotj['year'], sotj['total_positive_misalignment'], label='SOTJ Formula')
plt.plot(ccctb['year'], ccctb['total_positive_misalignment'], label='CCCTB Formula')
plt.plot(doubleweightedsales['year'], doubleweightedsales['total_positive_misalignment'], label='Double-Weighted Sales Formula')
plt.plot(canadian['year'], canadian['total_positive_misalignment'], label='Canadian Formula')
plt.plot(threefactor['year'], threefactor['total_positive_misalignment'], label='Three-Factor Formula')

plt.title('Total Misalignment Over the Years, Global Estimate')
plt.xlabel('Year')
plt.ylabel('Total Misalignment (in million USD)')
plt.ylim(0, )

# Set x ticks to show only full years
years = sorted(set(sotj['year']).union(set(ccctb['year'])).union(set(doubleweightedsales['year'])).union(set(canadian['year'])).union(set(threefactor['year'])))
plt.xticks(years)  # Set x ticks to the unique years

plt.legend()
plt.gca().yaxis.set_major_formatter(ticker.FuncFormatter(lambda x, pos: f'{x:,.0f}'))
plt.grid(True)
plt.show()
```
