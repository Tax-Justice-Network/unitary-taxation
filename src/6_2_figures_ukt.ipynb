```python
import matplotlib.pyplot as plt
from matplotlib import rcParams
import numpy as np
import os

# ---------------------------------
# Fixed values in your preferred order
# ---------------------------------
data = [
    ("Responsibility for\n corporate tax abuse", 23.2, "#50805E"),
    ("Responsibility for\n individual tax abuse", 33.1, "#FFD371"),
    ("\nPopulation", 0.9, "#AD756C"),
    ("\nGDP", 3.2, "#586AAD"),
]

# ---------------------------------
# Save folder
# ---------------------------------
outdir = r"C:\Users\aliso\Dropbox\Apps\Overleaf\The UK Spider Web (1)\1_figures"

# Create folder if needed
os.makedirs(outdir, exist_ok=True)

# ---------------------------------
# Fonts
# ---------------------------------
rcParams["font.family"] = "sans-serif"
rcParams["font.sans-serif"] = ["Work Sans", "Montserrat", "DejaVu Sans", "Arial"]

# ---------------------------------
# Colors / style
# ---------------------------------
sotj_dark_purple = "#564A5C"
bar_width = 0.58
ymax_all = max(v for _, v, _ in data)

# ---------------------------------
# Chart function
# ---------------------------------
def make_chart(step, filename):
    subset = data[:step]

    labels = [item[0] for item in subset]
    values = [item[1] for item in subset]
    colors = [item[2] for item in subset]

    # ---------------------------------
    # Two layout regimes:
    # steps 1-2 fixed together
    # steps 3-4 fixed together
    # ---------------------------------
    if step <= 2:
        # Use your preferred spacing for first two
        x = np.array([0, 2])[:step]
        xlim_left, xlim_right = -0.8, 2.8
        fig_w = 7.2
    else:
        # Use your preferred full layout for 3 and 4
        x = np.array([0, 2, 4, 5])[:step]
        xlim_left, xlim_right = -0.8, 5.8
        fig_w = 10

    fig, ax = plt.subplots(figsize=(fig_w, 6.5), facecolor="white")
    ax.set_facecolor("white")

    bars = ax.bar(x, values, color=colors, width=bar_width)

    # Remove all axes and ticks
    for spine in ["top", "right", "left", "bottom"]:
        ax.spines[spine].set_visible(False)

    ax.set_yticks([])
    ax.tick_params(axis="x", length=0, colors=sotj_dark_purple, pad=14)
    ax.tick_params(axis="y", length=0)

    # Category labels below
    ax.set_xticks(x)
    ax.set_xticklabels(labels, fontsize=18, color=sotj_dark_purple)

    # Fixed limits within each regime
    ax.set_ylim(0, ymax_all * 1.22)
    ax.set_xlim(xlim_left, xlim_right)

    # Value labels above bars
    for bar, val in zip(bars, values):
        ax.text(
            bar.get_x() + bar.get_width() / 2,
            val + ymax_all * 0.025,
            f"{val:.1f}%",
            ha="center",
            va="bottom",
            fontsize=15,
            color=sotj_dark_purple,
            fontweight="semibold"
        )
    ax.set_title(
    "UK and associated jurisdictions: Share of the global total",
    fontsize=16,
    color=sotj_dark_purple,
    pad=14,
    fontweight="semibold"
    )

    plt.tight_layout()

    # IMPORTANT: no bbox_inches="tight"
    plt.savefig(os.path.join(outdir, filename), dpi=300)
    plt.close()

# ---------------------------------
# Export step-by-step files
# ---------------------------------
make_chart(1, "uk_footprint_step1.pdf")
make_chart(2, "uk_footprint_step2.pdf")
make_chart(3, "uk_footprint_step3.pdf")
make_chart(4, "uk_footprint_step4.pdf")

# Optional final PNG
make_chart(4, "uk_footprint_final.png")
```

```python
import matplotlib.pyplot as plt
from matplotlib import rcParams
import os

# ---------------------------------
# Data (current USD, 2021, in bn)
# ---------------------------------
corp_loss = 80.521
indiv_loss = 48.962
africa_health = 61.6
total_loss = corp_loss + indiv_loss

# ---------------------------------
# Save folder
# ---------------------------------
outdir = r"C:\Users\aliso\Dropbox\Apps\Overleaf\The UK Spider Web (1)\1_figures"
os.makedirs(outdir, exist_ok=True)

# ---------------------------------
# Fonts
# ---------------------------------
rcParams["font.family"] = "sans-serif"
rcParams["font.sans-serif"] = ["Work Sans", "Montserrat", "DejaVu Sans", "Arial"]

# ---------------------------------
# Colors
# ---------------------------------
green = "#50805E"      # corporate
gold = "#FFD371"       # individual
blue = "#586AAD"       # comparator bar
dark = "#564A5C"

def fmt(x):
    return f"USD {x:.1f}bn"

def make_chart(step, filename):
    # step 1: corporate only
    # step 2: add individual stacked on top + total
    # step 3: add Africa comparator
    x_stacked = 0
    x_africa = 2.0
    bar_width = 0.72

    ymax = max(total_loss, africa_health) * 1.22

    fig, ax = plt.subplots(figsize=(9.2, 6.8), facecolor="white")
    ax.set_facecolor("white")

    # Stacked bar: corporate base always visible
    corp_bar = ax.bar(
        x_stacked, corp_loss, width=bar_width, color=green
    )[0]

    # Individual appears on top from step 2 onward
    if step >= 2:
        indiv_bar = ax.bar(
            x_stacked, indiv_loss, width=bar_width, bottom=corp_loss, color=gold
        )[0]
    else:
        indiv_bar = None

    # Africa comparator appears only from step 3 onward
    if step >= 3:
        africa_bar = ax.bar(
            x_africa, africa_health, width=bar_width, color=blue
        )[0]
    else:
        africa_bar = None

    # Clean axes
    for spine in ["top", "right", "left", "bottom"]:
        ax.spines[spine].set_visible(False)

    ax.set_yticks([])
    ax.tick_params(axis="x", length=0, colors=dark, pad=14)
    ax.tick_params(axis="y", length=0)

    # Bottom labels
    if step >= 3:
        xticks = [x_stacked, x_africa]
        xlabels = [
            "Revenue losses caused\nby UK-linked jurisdictions",
            "Government health expenditure\nof all African countries combined"
        ]
    else:
        xticks = [x_stacked]
        xlabels = ["Revenue losses caused\nby UK-linked jurisdictions"]

    ax.set_xticks(xticks)
    ax.set_xticklabels(xlabels, fontsize=16, color=dark)

    ax.set_xlim(-0.9, 3.1)
    ax.set_ylim(0, ymax)

    # Corporate value label inside green segment
    ax.text(
        x_stacked,
        corp_loss / 2,
        fmt(corp_loss),
        ha="center",
        va="center",
        fontsize=15,
        color="white",
        fontweight="semibold"
    )

    # Corporate segment name to the right
    ax.text(
        x_stacked + bar_width / 2 + 0.18,
        corp_loss / 2,
        "Corporate\ntax abuse",
        ha="left",
        va="center",
        fontsize=15,
        color=dark
    )

    # Individual segment label + name + total label once it appears
    if step >= 2:
        ax.text(
            x_stacked,
            corp_loss + indiv_loss / 2,
            fmt(indiv_loss),
            ha="center",
            va="center",
            fontsize=15,
            color=dark,
            fontweight="semibold"
        )

        ax.text(
            x_stacked + bar_width / 2 + 0.18,
            corp_loss + indiv_loss / 2,
            "Individual\ntax abuse",
            ha="left",
            va="center",
            fontsize=15,
            color=dark
        )

        ax.text(
            x_stacked,
            total_loss + ymax * 0.02,
            f"Total: {fmt(total_loss)}",
            ha="center",
            va="bottom",
            fontsize=15,
            color=dark,
            fontweight="semibold"
        )

    # Africa bar label
    if step >= 3:
        ax.text(
            x_africa,
            africa_health + ymax * 0.018,
            fmt(africa_health),
            ha="center",
            va="bottom",
            fontsize=15,
            color=dark,
            fontweight="semibold"
        )

    plt.tight_layout()
    plt.savefig(os.path.join(outdir, filename), dpi=300)
    plt.close()

# ---------------------------------
# Export files
# ---------------------------------
make_chart(1, "uk_stacked_losses_vs_africa_health_step1.pdf")
make_chart(2, "uk_stacked_losses_vs_africa_health_step2.pdf")
make_chart(3, "uk_stacked_losses_vs_africa_health_step3.pdf")
make_chart(3, "uk_stacked_losses_vs_africa_health_final.png")
```

```python
import matplotlib.pyplot as plt
from matplotlib import rcParams
import numpy as np
import os

# ---------------------------------
# Data (current mUSD, 2021)
# ---------------------------------
gain = 4097    # tax revenue gained
loss = 16901   # tax revenue loss suffered

# ---------------------------------
# Save folder
# ---------------------------------
outdir = r"C:\Users\aliso\Dropbox\Apps\Overleaf\The UK Spider Web (1)\1_figures"
os.makedirs(outdir, exist_ok=True)

# ---------------------------------
# Fonts
# ---------------------------------
rcParams["font.family"] = "sans-serif"
rcParams["font.sans-serif"] = ["Work Sans", "Montserrat", "DejaVu Sans", "Arial"]

# ---------------------------------
# Colors
# ---------------------------------
green_gain = "#50805E"
green_loss = "#AD756C"
dark = "#564A5C"

def fmt_bn(x):
    return f"{x/1000:.1f}bn USD"

# Fixed positions so both overlays keep the same framing
x_all = np.array([0.0, 3.6])
bar_width = 0.72
ymax_all = max(gain, loss) * 1.22

def make_chart(step, filename):
    labels = [
        "Tax revenue UK-linked jurisdictions\ngain from stealing taxing rights",
        "Tax revenue UK-linked jurisdictions\nlose from corporate tax abuse"
    ]
    values = [gain, loss]
    colors = [green_gain, green_loss]

    fig, ax = plt.subplots(figsize=(10.2, 6.8), facecolor="white")
    ax.set_facecolor("white")

    # Draw only the bars that should be visible
    if step >= 1:
        ax.bar(x_all[0], values[0], color=colors[0], width=bar_width)
    if step >= 2:
        ax.bar(x_all[1], values[1], color=colors[1], width=bar_width)

    # Remove axes
    for spine in ["top", "right", "left", "bottom"]:
        ax.spines[spine].set_visible(False)

    ax.set_yticks([])
    ax.tick_params(axis="x", length=0, colors=dark, pad=16)
    ax.tick_params(axis="y", length=0)

    # Keep both tick positions fixed, but only reveal labels when needed
    shown_labels = [
        labels[0] if step >= 1 else "",
        labels[1] if step >= 2 else ""
    ]
    ax.set_xticks(x_all)
    ax.set_xticklabels(shown_labels, fontsize=16, color=dark)

    ax.set_xlim(-1.1, 4.9)
    ax.set_ylim(0, ymax_all)

    # Value labels above bars
    if step >= 1:
        ax.text(
            x_all[0],
            values[0] + ymax_all * 0.02,
            fmt_bn(values[0]),
            ha="center",
            va="bottom",
            fontsize=15,
            color=dark,
            fontweight="semibold"
        )

    if step >= 2:
        ax.text(
            x_all[1],
            values[1] + ymax_all * 0.02,
            fmt_bn(values[1]),
            ha="center",
            va="bottom",
            fontsize=15,
            color=dark,
            fontweight="semibold"
        )

    plt.tight_layout()
    plt.savefig(os.path.join(outdir, filename), dpi=300)
    plt.close()

make_chart(1, "uk_payoff_step1.pdf")
make_chart(2, "uk_payoff_step2.pdf")
make_chart(2, "uk_payoff_final.png")
```
