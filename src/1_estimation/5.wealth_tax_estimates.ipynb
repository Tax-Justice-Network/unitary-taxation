# Estimate wealth tax revenues for EU Member States

```python
# import packages
import pandas as pd
import tjn_tools
import numpy as np
import tjn_internal
#from config import WEALTH_TAX, WEALTH_TAX_INPUT, WEALTH_TAX_OUTPUT
WEALTH_TAX = f"{tjn_internal.paths.sharepoint_root}/TJN - Shared Documents/Workstreams/Scale of Tax Injustice/Wealth tax/2_data"
WEALTH_TAX_INPUT = f"{WEALTH_TAX}/1_input"
WEALTH_TAX_OUTPUT = f"{WEALTH_TAX}/2_output"
```

### Create dataset

```python
top1 = pd.read_csv(f"{WEALTH_TAX_INPUT}/WID_top1.csv", sep=';', header=1)
top1 = top1.iloc[:, [0, 2, 3]].rename(columns={top1.columns[0]: 'country',top1.columns[2]: 'twealth_top1',top1.columns[3]: 'awealth_top1'})
top05 = pd.read_csv(f"{WEALTH_TAX_INPUT}/WID_top05.csv", sep=';', header=1)
top05 = top05.iloc[:, [0, 2, 3]].rename(columns={top05.columns[0]: 'country',top05.columns[2]: 'twealth_top05',top05.columns[3]: 'awealth_top05'})
top01 = pd.read_csv(f"{WEALTH_TAX_INPUT}/WID_top01.csv", sep=';', header=1)
top01 = top01.iloc[:, [0, 2, 3]].rename(columns={top01.columns[0]: 'country',top01.columns[2]: 'twealth_top01',top01.columns[3]: 'awealth_top01'})
top005 = pd.read_csv(f"{WEALTH_TAX_INPUT}/WID_top005.csv", sep=';', header=1)
top005 = top005.iloc[:, [0, 2, 3]].rename(columns={top005.columns[0]: 'country',top005.columns[2]: 'twealth_top005',top005.columns[3]: 'awealth_top005'})
top001 = pd.read_csv(f"{WEALTH_TAX_INPUT}/WID_top001.csv", sep=';', header=1)
top001 = top001.iloc[:, [0, 2, 3]].rename(columns={top001.columns[0]: 'country',top001.columns[2]: 'twealth_top001',top001.columns[3]: 'awealth_top001'})
pop = pd.read_csv(f"{WEALTH_TAX_INPUT}/WID_population.csv", sep=';', header=1)
pop = pop.iloc[:, [0, 2]].rename(columns={pop.columns[0]: 'country',pop.columns[2]: 'adult_population'})
wealth_data = pop.merge(top1, on='country').merge(top05, on='country').merge(top01, on='country').merge(top005, on='country').merge(top001, on='country')
wealth_data.to_excel(f"{WEALTH_TAX_OUTPUT}/Wealth_tax_EU.xlsx", index=None)
```
